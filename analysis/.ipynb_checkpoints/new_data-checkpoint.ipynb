```python
df = None 

for each in files:
    if df is None:
        df = pd.read_csv(path+each)
        continue
    temp = pd.read_csv(path+each)
    if temp.empty or temp.isna().all().all():
        print(path+each)
    df = pd.concat([df, temp], ignore_index=True) 
```
