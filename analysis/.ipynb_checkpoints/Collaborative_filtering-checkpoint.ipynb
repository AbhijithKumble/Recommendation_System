```python
import pandas as pd
import numpy as np
import gc # For manual garbage collection
import io # For dummy data example

def calculate_anime_anime_similarity_ultra_large(chunk_source, total_anime_ids, total_user_ids):
    """
    Calculates anime-anime cosine similarity incrementally for extremely large chunks.
    Assumes each chunk can be loaded, but the A @ A.T operation within it causes crashes.

    Args:
        chunk_source (list or generator): A list of Pandas DataFrames, or a generator that yields DataFrame chunks.
                                         Each DataFrame should have anime_id as index
                                         and user_id as columns.
        total_anime_ids (list): A sorted list of all unique anime IDs across all chunks.
                                This defines the rows/columns of the output similarity matrix.
        total_user_ids (list): A sorted list of all unique user IDs across all chunks.
                                Used to ensure consistent column order in chunks.

    Returns:
        pandas.DataFrame: The anime-anime cosine similarity matrix.
    """
    num_total_anime = len(total_anime_ids)
    num_total_users = len(total_user_ids) # Needed for reindexing/alignment

    # Initialize accumulators for dot products and squared norms
    dot_products_sum = np.zeros((num_total_anime, num_total_anime))
    norms_squared_sum = np.zeros(num_total_anime)

    # Create mappings for efficient index lookup
    anime_id_to_idx = {anime_id: i for i, anime_id in enumerate(total_anime_ids)}
    
    # Iterate through chunks (e.g., provided as a list of DataFrames or from files)
    # If a single chunk file cannot even be loaded by pd.read_csv, this needs further adaptation
    # (e.g., using Dask or iterating rows with pd.read_csv(chunksize=...))
    for i, chunk_df in enumerate(chunk_source):
        print(f"Processing chunk {i+1}...")

        # Ensure chunk_df has all users as columns and in the correct order
        # Fill missing user ratings for anime with 0 (or appropriate default)
        chunk_df_aligned = chunk_df.reindex(columns=total_user_ids, fill_value=0.0)
        
        # Convert to NumPy array for efficient numeric operations
        # .values.T transposes the chunk, so rows are users, columns are anime in this chunk
        # Shape: (num_users_total, num_anime_in_chunk)
        user_ratings_for_this_chunk_anime = chunk_df_aligned.values.T.astype(float)
        
        # Get global indices for anime present in this chunk
        current_chunk_anime_ids = chunk_df.index.tolist()
        current_chunk_anime_global_indices = [anime_id_to_idx[aid] for aid in current_chunk_anime_ids]

        # --- Incremental Calculation by User (Column) within the Chunk ---
        # This avoids creating the large (num_anime_in_chunk x num_anime_in_chunk) intermediate matrix

        for user_idx in range(num_total_users):
            # Extract one user's ratings for the anime in this chunk
            # Shape: (num_anime_in_chunk,)
            user_vector_for_chunk_anime = user_ratings_for_this_chunk_anime[user_idx, :]

            # Update norms_squared_sum for each anime
            squared_ratings_for_user = user_vector_for_chunk_anime**2
            for local_anime_idx, global_anime_idx in enumerate(current_chunk_anime_global_indices):
                norms_squared_sum[global_anime_idx] += squared_ratings_for_user[local_anime_idx]

            # Update dot_products_sum
            # Compute outer product for this user's contribution
            # Resulting matrix is (num_anime_in_chunk x num_anime_in_chunk)
            # This matrix is added to the global sum (element-wise), avoiding a single large temp matrix
            outer_product_for_user = np.outer(user_vector_for_chunk_anime, user_vector_for_chunk_anime)
            
            # Add to the global dot_products_sum based on global anime indices
            for r_local, r_global in enumerate(current_chunk_anime_global_indices):
                for c_local, c_global in enumerate(current_chunk_anime_global_indices):
                    dot_products_sum[r_global, c_global] += outer_product_for_user[r_local, c_local]

        # Clear chunk_df from memory
        del chunk_df
        del chunk_df_aligned
        del user_ratings_for_this_chunk_anime
        gc.collect()

    # --- Final Cosine Similarity Calculation ---
    similarity_matrix = np.zeros((num_total_anime, num_total_anime))
    for i in range(num_total_anime):
        for j in range(num_total_anime):
            # Avoid division by zero
            denom_i = np.sqrt(norms_squared_sum[i])
            denom_j = np.sqrt(norms_squared_sum[j])
            
            if denom_i > 0 and denom_j > 0:
                similarity_matrix[i, j] = dot_products_sum[i, j] / (denom_i * denom_j)
            else:
                similarity_matrix[i, j] = 0.0 # Or NaN, if you prefer

    # Convert to DataFrame for better readability
    cosine_sim_df = pd.DataFrame(similarity_matrix, index=total_anime_ids, columns=total_anime_ids)
    return cosine_sim_df
```
