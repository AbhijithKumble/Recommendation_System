```python
top_similar_items_dict = {}
top_similar_how_much = 10
entries = list(zip(sim_matrix_index, sim_matrix))

for each in entries:
    _anime_id = each[0]
    _anime_list = each[1]
    similarities = list(zip(sim_matrix_index,_anime_list))
    similarities.sort(key=lambda x: x[1], reverse=True)
    top_similar_items_dict[_anime_id]=similarities[1:11]
```

```python
def get_recommendation(sim_dict, user_history,top_n=10):
    '''
        sim_dict: Dict[int, List[(int, np.float),]]
        user_history: set[int]
    '''
    users_seen = set(user_history)
    recs = set()
    score=0 
    for anime_id in users_seen:
        similar_animes = sim_dict.get(anime_id, [])
        similar_animes = similar_animes[:top_n]
        for sim_anime_id in similar_animes:
            if sim_anime_id[0] in users_seen:
                recs.add(sim_anime_id[0])
        output = len(recs.intersection(users_seen))/top_n
        recs.clear()
        score+=output
    print(score/len(users_seen))
```

```python
user_history = merged_df.groupby('username')['anime_id'].apply(set)
```

```python
relevant = get_recommendation(top_similar_items_dict,user_history.iloc[0])
```

```python
LIKED_SCORE_THRESHOLD = 7

positive_interactions_df = merged_df[merged_df['my_score'] >= LIKED_SCORE_THRESHOLD]

user_positive_anime = positive_interactions_df.groupby('user_id')['anime_id'].apply(set).to_dict()

K = 10 
all_precisions_at_K = []

for source_anime_id, recommended_anime_tuples in top_similar_items_dict.items():
    recommended_anime_ids = [anime_id for anime_id, _ in recommended_anime_tuples[:K]]

    users_who_liked_source = positive_interactions_df[
        positive_interactions_df['anime_id'] == source_anime_id
    ]['user_id'].unique()

    if len(users_who_liked_source) == 0:
        continue # Skip if no users have liked this source anime (cannot evaluate for it)

    source_anime_precisions = []
    for user_id in users_who_liked_source:
        relevant_items_for_user = user_positive_anime.get(user_id, set())

        if not relevant_items_for_user:
            continue # User has no liked items, cannot evaluate

        hits = 0
        for rec_anime_id in recommended_anime_ids:
            if rec_anime_id in relevant_items_for_user:
                hits += 1
        
        denominator = min(K, len(recommended_anime_ids))
        if denominator > 0:
            user_precision_for_source = hits / denominator
            source_anime_precisions.append(user_precision_for_source)
    
    if source_anime_precisions:
        all_precisions_at_K.extend(source_anime_precisions) # Add individual user precisions

overall_precision_at_K = sum(all_precisions_at_K) / len(all_precisions_at_K) if all_precisions_at_K else 0

print(f"Overall Precision@{K} for item-item similarity: {overall_precision_at_K:.4f}")


all_recalls_at_K = []

for source_anime_id, recommended_anime_tuples in top_similar_items_dict.items():
    recommended_anime_ids = [anime_id for anime_id, _ in recommended_anime_tuples[:K]]
    users_who_liked_source = positive_interactions_df[
        positive_interactions_df['anime_id'] == source_anime_id
    ]['user_id'].unique()

    if len(users_who_liked_source) == 0:
        continue

    for user_id in users_who_liked_source:
        relevant_items_for_user = user_positive_anime.get(user_id, set())
        
        # Exclude the source_anime_id itself from relevant items if evaluating for future items
        # Or keep it if you're evaluating if other liked items are similar
        # For simplicity, here we consider all relevant items for the user
        
        if not relevant_items_for_user:
            continue

        hits = 0
        for rec_anime_id in recommended_anime_ids:
            if rec_anime_id in relevant_items_for_user:
                hits += 1
      
        relevant_items_excluding_source = relevant_items_for_user - {source_anime_id}

        if len(relevant_items_excluding_source) > 0:
            recall = hits / len(relevant_items_excluding_source)
            all_recalls_at_K.append(recall)

overall_recall_at_K = sum(all_recalls_at_K) / len(all_recalls_at_K) if all_recalls_at_K else 0
print(f"Overall Recall@{K} for item-item similarity: {overall_recall_at_K:.4f}")

# F1-score
if overall_precision_at_K + overall_recall_at_K > 0:
    overall_f1_score = 2 * (overall_precision_at_K * overall_recall_at_K) / (overall_precision_at_K + overall_recall_at_K)
    print(f"Overall F1-score@{K}: {overall_f1_score:.4f}")
else:
    print(f"Overall F1-score@{K}: 0.0000 (Precision and Recall are zero)")
```
