```python
user_anime_matrix = []
for i in range(len(cleaned_chunks)):
    user_anime_matrix.append(cleaned_chunks[i].pivot_table(index='anime_id', columns='user_id', values='my_score').fillna(0))
```

```python
user_ids_set = []
anime_ids_set = set(list(user_anime_matrix[0].index))

for i in range(len(user_anime_matrix)):
    col = list(user_anime_matrix[i].columns)
    user_ids_set+=col

print(len(user_ids_set), len(anime_ids_set))
```

```python
user_anime_matrix[0]
```

```python
try:
    combined_df = pd.concat(user_anime_matrix, axis=1)
except MemoryError:
    print("❌ Not enough memory to combine DataFrames!")
```
